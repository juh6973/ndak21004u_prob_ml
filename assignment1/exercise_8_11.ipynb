# Exercise 8.11: The fuel gauge and the unreliable driver

Battery $B$ and fuel tank $F$ (1 = ok, 0 = flat/empty) drive the gauge $G$ (1 = full, 0 = empty). A driver $D$ reads the gauge and reports it, sometimes wrongly: $B \to G \leftarrow F$, $G \to D$.

$$
\begin{aligned}
p(B=1) &= p(F=1) = 0.9 \quad \text{(independent)}\\
p(G=1 \mid B=1, F=1) &= 0.8, \qquad p(G=1 \mid B=1, F=0) = 0.2\\
p(G=1 \mid B=0, F=1) &= 0.2, \qquad p(G=1 \mid B=0, F=0) = 0.1\\
p(D=1 \mid G=1) &= p(D=0 \mid G=0) = 0.9
\end{aligned}
$$

$$p(B, F, G, D) = p(B)\,p(F)\,p(G \mid B, F)\,p(D \mid G)$$

**Shortcut.** $D$ only depends on $G$, so for any condition "$\cdot$":

$$p(D=0 \mid \cdot) = 0.9\, p(G=0 \mid \cdot) + 0.1\,\big(1 - p(G=0 \mid \cdot)\big) = 0.1 + 0.8\, p(G=0 \mid \cdot)$$

## Part 1: $p(F=0 \mid D=0)$

Average over the battery to get the gauge, then use the shortcut for the report:

$$
\begin{aligned}
p(G=0 \mid F=0) &= 0.81, & p(D=0 \mid F=0) &= 0.1 + 0.8 \cdot 0.81 = 0.748\\
p(G=0 \mid F=1) &= 0.2 \cdot 0.9 + 0.8 \cdot 0.1 = 0.26, & p(D=0 \mid F=1) &= 0.1 + 0.8 \cdot 0.26 = 0.308
\end{aligned}
$$

$$
p(F=0 \mid D=0) = \frac{0.748 \cdot 0.1}{0.748 \cdot 0.1 + 0.308 \cdot 0.9} = \frac{0.0748}{0.352} \approx \mathbf{0.213}
$$

Up from the prior 0.1, but below the 0.257 we'd get reading the gauge ourselves (book eq. 8.32): a report from an unreliable driver is weaker evidence.

## Part 2: $p(F=0 \mid D=0, B=0)$

$B$ and $F$ are independent, so $p(F \mid B=0) = p(F)$. With a flat battery:

$$
\begin{aligned}
p(D=0 \mid B=0, F=0) &= 0.1 + 0.8 \cdot 0.9 = 0.82\\
p(D=0 \mid B=0, F=1) &= 0.1 + 0.8 \cdot 0.8 = 0.74
\end{aligned}
$$

$$
p(F=0 \mid D=0, B=0) = \frac{0.82 \cdot 0.1}{0.82 \cdot 0.1 + 0.74 \cdot 0.9} = \frac{0.082}{0.748} \approx \mathbf{0.110}
$$

## Intuition and Figure 8.54

**Explaining away.** An "empty" report has two possible causes: an empty tank or a flat battery. Knowing the battery is flat already explains the report, so the tank gets less blame and its probability drops back towards the prior 0.1. The book shows the same drop for the gauge itself ($0.257 \to 0.111$); the driver's noise pulls both numbers closer to the prior.

**Figure 8.54** is our graph with $a = B$, $b = F$, $c = G$, $d = D$. With nothing observed, the path $B \to G \leftarrow F$ is blocked and $B$, $F$ are independent. Observing $G$, or a descendant of $G$ such as $D$, opens the path and makes $B$ and $F$ dependent. That is why learning $B$ changes our belief about $F$ in Part 2. Since $D$ is only a noisy copy of $G$, the link is weaker than when observing $G$ directly.
